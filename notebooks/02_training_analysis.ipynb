# NeRF Training Analysis

This notebook analyzes training progress, visualizes loss curves, and compares rendered vs. ground truth images.

```python
import sys
from pathlib import Path
import torch
import matplotlib.pyplot as plt
import numpy as np
from tensorboard.backend.event_processing import event_accumulator

sys.path.insert(0, str(Path('..').resolve()))
```

## 1. Load Training Logs from TensorBoard

```python
# Path to TensorBoard logs
LOG_DIR = '../outputs/logs'  # Update with your log directory

# Load events
ea = event_accumulator.EventAccumulator(LOG_DIR)
ea.Reload()

# Get available scalar keys
print("Available metrics:")
for key in ea.Tags()['scalars']:
    print(f"  - {key}")
```

## 2. Plot Loss Curves

```python
# Extract loss data
total_loss = ea.Scalars('train/total_loss')
fine_loss = ea.Scalars('train/fine_loss')
coarse_loss = ea.Scalars('train/coarse_loss')

# Plot
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

# Total loss
steps = [s.step for s in total_loss]
values = [s.value for s in total_loss]
axes[0].plot(steps, values, label='Total Loss')
axes[0].set_xlabel('Training Step')
axes[0].set_ylabel('Loss')
axes[0].set_title('Training Loss')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Fine and coarse losses
fine_steps = [s.step for s in fine_loss]
fine_values = [s.value for s in fine_loss]
coarse_steps = [s.step for s in coarse_loss]
coarse_values = [s.value for s in coarse_loss]

axes[1].plot(fine_steps, fine_values, label='Fine Loss')
axes[1].plot(coarse_steps, coarse_values, label='Coarse Loss')
axes[1].set_xlabel('Training Step')
axes[1].set_ylabel('Loss')
axes[1].set_title('Coarse vs Fine Loss')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 3. Plot Learning Rate Schedule

```python
# Extract learning rate
lr_data = ea.Scalars('train/lr')
lr_steps = [s.step for s in lr_data]
lr_values = [s.value for s in lr_data]

plt.figure(figsize=(10, 6))
plt.plot(lr_steps, lr_values)
plt.xlabel('Training Step')
plt.ylabel('Learning Rate')
plt.title('Learning Rate Schedule')
plt.yscale('log')
plt.grid(True, alpha=0.3)
plt.show()
```
